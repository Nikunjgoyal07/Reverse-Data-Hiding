```python
import os
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import torchvision.transforms as transforms
from PIL import Image
import numpy as np

# Custom Dataset class for preprocessed Cross and Dot images
class CrossDotDataset(Dataset):
    def __init__(self, cross_dir, dot_dir):
        self.cross_dir = cross_dir
        self.dot_dir = dot_dir
        
        # Get list of image files
        self.cross_files = sorted([f for f in os.listdir(cross_dir) if f.endswith('.png')])
        self.dot_files = sorted([f for f in os.listdir(dot_dir) if f.endswith('.png')])
        
        # Verify both directories have the same number of files
        assert len(self.cross_files) == len(self.dot_files), "Number of cross and dot images must match"
        
    def __len__(self):
        return len(self.cross_files)
    
    def __getitem__(self, idx):
        # Load cross image
        cross_path = os.path.join(self.cross_dir, self.cross_files[idx])
        cross_img = Image.open(cross_path)
        
        # Load dot image  
        dot_path = os.path.join(self.dot_dir, self.dot_files[idx])
        dot_img = Image.open(dot_path)
        
        # Convert to tensors (assuming images are already 512x512 grayscale)
        transform = transforms.ToTensor()
        cross_tensor = transform(cross_img)
        dot_tensor = transform(dot_img)
        
        return cross_tensor, dot_tensor

# Initialize dataset
cross_directory = "dataset_cross"    # Update this path
dot_directory = "dataset_dot"        # Update this path

dataset = CrossDotDataset(cross_dir=cross_directory, dot_dir=dot_directory)

# Create data loader
data_loader = DataLoader(dataset, batch_size=4, shuffle=True)

# Print dataset info
print(f"Dataset size: {len(dataset)} images")
print(f"Number of batches: {len(data_loader)}")

# Test one batch to verify everything works
for cross_batch, dot_batch in data_loader:
    print(f"Cross batch shape: {cross_batch.shape}")  # Should be [4, 1, H, W]
    print(f"Dot batch shape: {dot_batch.shape}")      # Should be [4, 1, H, W]
    print(f"Data type: {cross_batch.dtype}")
    print(f"Cross pixel range: [{cross_batch.min():.3f}, {cross_batch.max():.3f}]")
    print(f"Dot pixel range: [{dot_batch.min():.3f}, {dot_batch.max():.3f}]")
    break
```
